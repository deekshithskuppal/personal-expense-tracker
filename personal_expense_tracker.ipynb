```python

expenses = []

while True:
    amount = float(input("Enter expense amount: ₹"))
    category = input("Enter category: ")
    description = input("Enter description: ")

    expense = {
        "amount": amount,
        "category": category,
        "description": description
    }

    expenses.append(expense)

    choice = input("Add another expense? (yes/no): ")

    if choice.lower() != "yes":
        break

total = sum(expense["amount"] for expense in expenses)

print("\n--- Expense Report ---")

for expense in expenses:
    print(expense)

print("Total spending: ₹", total)
```

```python

from datetime import date

expenses = []

while True:
    amount = float(input("Enter expense amount: ₹"))
    category = input("Enter category: ")
    description = input("Enter description: ")

    expense = {
        "date": str(date.today()),
        "amount": amount,
        "category": category,
        "description": description
    }

    expenses.append(expense)

    choice = input("Add another expense? (yes/no): ")

    if choice.lower() != "yes":
        break

total = sum(item["amount"] for item in expenses)

print("\n--- Expense Report ---")

for item in expenses:
    print(item["date"], "|", item["category"],
          "| ₹", item["amount"], "|", item["description"])

print("\nTotal spending: ₹", total)
```

```python

import csv
from datetime import date

expenses = []

while True:
    amount = float(input("Enter expense amount: ₹"))
    category = input("Enter category: ")
    description = input("Enter description: ")

    expense = {
        "date": str(date.today()),
        "amount": amount,
        "category": category,
        "description": description
    }

    expenses.append(expense)

    choice = input("Add another expense? (yes/no): ")

    if choice.lower() != "yes":
        break

with open("expenses.csv", "w", newline="") as file:
    fields = ["date", "amount", "category", "description"]
    writer = csv.DictWriter(file, fieldnames=fields)

    writer.writeheader()
    writer.writerows(expenses)

print("Expenses saved successfully!")

total = sum(item["amount"] for item in expenses)
print("Total spending: ₹", total)
```

```python

import pandas as pd
import matplotlib.pyplot as plt

# Read the saved expense file
df = pd.read_csv("expenses.csv")

# Show all expenses
print("My Expense Records:")
display(df)

# Calculate total spending
total = df["amount"].sum()
print("Total Spending: ₹", total)

# Calculate spending by category
category_total = df.groupby("category")["amount"].sum()

print("\nCategory-wise Spending:")
display(category_total)

# Draw a chart
category_total.plot(kind="bar")
plt.title("Expense Tracker - Category-wise Spending")
plt.xlabel("Category")
plt.ylabel("Amount (₹)")
plt.show()
```

```python

import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("expenses.csv")

print("💰 PERSONAL EXPENSE TRACKER")
print("=" * 35)

print("\nAll Expense Records:")
display(df)

total = df["amount"].sum()
print(f"\nTotal Spending: ₹{total:.2f}")

print("\nCategory-wise Spending:")
category_total = df.groupby("category")["amount"].sum()

for category, amount in category_total.items():
    print(f"{category}: ₹{amount:.2f}")

category_total.plot(kind="bar", title="Spending by Category")
plt.ylabel("Amount (₹)")
plt.xlabel("Category")
plt.tight_layout()
plt.show()
```
